# Review of loops over strings

Many problems that involve making a loop over a string involve either **counting** something inside the string
or **building a new string** from an existing string.

## Patterns for these ideas:

__Generic counting function:__
```python
def counting_function(string):
    count = 0
    for idx in range(0, len(string)):
        if <some condition>:
            count += 1
    return count
```

__Generic string-building function:__
```python
def building_function(string):
    answer = ""
    for idx in range(0, len(string)):
        if <some condition>:
            answer = answer + string[idx]
        # maybe more conditions here...
    return answer
```

Notice how both use the same `for` loop and a variable that keeps either a running total (counting) or a string variable that successively builds the new string
starting from the empty string `""`.

## Counting

Last week we wrote code that counts the number of uppercase characters in a string.  Here it is as a function:

````python
def count_upper(string):
    count = 0
    
    for idx in range(0, len(string)):
        ch = string[idx]  # make a variable to hold the current character
        if ch.isupper():
            count += 1
    
    return count

# Test cases for this function:
print(count_upper("BanANA"))  
print(count_upper("abracadabra"))  
print(count_upper("Computer Science 141 is awesome!"))  
````

Or, done a different way:

````python
def count_upper(string):
    count = 0
    
    for idx in range(0, len(string)):
        if string[idx].isupper():  # test current character directly
            count += 1
    
    return count

# Test cases for this function:
print(count_upper("BanANA"))  
print(count_upper("abracadabra"))  
print(count_upper("Computer Science 141 is awesome!")) 
````

## Building a new string

Many problems involving strings require us to build a new string out of an old one.  This is often done using string concatenation inside of a `for` loop.

Example: Above we have a function that counts the number of capital letters in a string.  Suppose we wanted to build a function that **removed** anything from the string that was not a capital letter.  In other words, the string "BanANA" would be turned into just "BANA".  

````python
def keep_capitals(string):
    answer = ""
    
    for idx in range(0, len(string)):
        if string[idx].isupper():
            answer = answer + string[idx]
    
    return answer

# Test cases for this function:
print(keep_capitals("BanANA"))  
print(keep_capitals("abracadabra"))  
print(keep_capitals("Computer Science 141 is awesome!")) 
````

## Another example

The code above is often called a **filter** because it lets certain characters pass through the `if` statement (capitals) but others don't (non-capitals).  So the answer
string consists of only certain characters from the original string.

Another common operation that uses this string building technique is sometimes called **transforming** a string.  Here, we turn every character of the original string
into one or more new characters.

This example doubles every character, so `"BanANA"` becomes `"BBaannAANNAA"`:

````python
def double_chars(string):
    answer = ""
    
    for idx in range(0, len(string)):
        answer = answer + string[idx] + string[idx]
    
    return answer

print(double_chars("BanANA"))  
print(double_chars("abracadabra"))  
print(double_chars("Computer Science 141 is awesome!")) 
````
