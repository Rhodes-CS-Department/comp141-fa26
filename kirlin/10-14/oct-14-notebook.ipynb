# File reading

* Programs use **files** to store data that they need while running.

* For example, word processors must load and save your document files.

* Your phone stores its contacts list in some sort of file.  

* Loading information from a file is called **reading from a file**.  
   
* We will be using **text files** for our reading and writing in this class.  That means the information in
the file is stored as plain numbers or words, rather than a binary format (that is harder to read).

## Reading from a file

* Python has many ways to read from a file.
* One of the most useful I call the **magic `for` loop**.


````python
# Example with file reading.

def main():
    file = open("supervillains.txt", "r")  # open the file for reading
    
    # loop over each line in the file, and assign the line
    # to the string variable called line.
    for line in file:
        print(line)
    
    file.close()   # close the file
    
main()
````

### Always three pieces to this process:
- open the file
- loop over the file
- close the file

## What does this `for line in file` thing do?

* A `for` loop is not only used for counting, it can also be used to **"loop over"** a file.  
    * That means the loop will read the file **one line at a time**, and the body of the loop will run once per line
    in the file.
    
    
* Each line of the file will be assigned **as a string** to the variable called `line`.
    * Like any counter variable in a `for` loop, the name of the `line` variable is not special.  You can call it
    anything you want.

Why are we getting extra blank lines?

Open the supervillain file and take a look at it.  Imagine you are typing in a new supervillain's name.  When
you get to the end of the line, you have to press enter to go to the next line.  The file stores those enter keypresses
(called newlines) and reads them back in when we read the file.  So when you  `print` each line of a file, we are seeing the 
newlines from the
file itself, **plus** the newline that the `print` statement always prints at the end of a line.


````python
# Example with file reading.

def main():
    file = open("supervillains.txt", "r")  
    
    for line in file:
        line = line.rstrip()   # "strip" the line, meaning remove whitespace from the right side
        print(line)
    
    # close the file
    file.close()
    
main()
````

## Reading numbers from a file

* When reading information from a file, the information is always read in as a **string**, even if the information in the file
is a number.

* Therefore, we must use the `int()` or `float()` functions to convert the information into integers or decimal numbers as
necessary after we read them from the file.

* This is actually the same way `input()` statements work, and why we have to use `int(input(...` or `float(input(...` to input
numbers from the keyboard.



````python
# Example with reading numbers.

def main():
    file = open("prices.txt", "r") 
    
    for line in file:
        line = line.rstrip()   
        print(line)
    
    file.close()
    
main()
````

````python
# Code along with me to make the example below read in prices and print
# the grand total at the end.

def main():
    file = open("prices.txt", "r")  
    
    total = 0
    for line in file:
        line = line.rstrip()   
        print(line)
    
    file.close()
    
    # print the grand total
    print("Your groceries cost", total)
    
main()
````

## Three steps for file reading and writing

1. Open the file.
2. Do the reading or writing.
3. Close the file.



````python
# Example with writing

def main():
    outfile = open("superheroes.txt", "w")  # open the file for writing
    
    # write some strings to the file
    outfile.write("Wonder Woman")
    outfile.write("Batman")
    outfile.write("Black Panther")
    outfile.write("The Hulk")
    outfile.write("Black Widow")
    
    # close the file
    outfile.close()
    
main()
    
````

Notice that when you run the code above, it will look like nothing happened.

However, a new file has been created called `superheroes.txt`.

But what's wrong with the file?

````python
# Modify this example to add newlines.

def main():
    outfile = open("superheroes.txt", "w")  # open the file for writing
    
    # write some strings to the file
    outfile.write("Wonder Woman")
    outfile.write("Batman")
    outfile.write("Black Panther")
    outfile.write("The Hulk")
    outfile.write("Black Widow")
    
    # close the file
    outfile.close()
    
main()
````

## How does this work?

* The `open` function opens a file either for reading or for writing, but not both.
    * You specify what you want with the second argument: either the string `"w"` for writing, or `"r"` for reading.
    * It returns a special object called a **file object** (that's the `outfile` thing in the code). 
    * You use this file object later to write to the file (put information into it).
    
    
    
* The `write` function writes information to a file.  It works mostly like the `print` function, except:
    * What you write **must** be a string (so if you want to store numbers in a file, you must convert them to
    strings first, and
    * You can only write one thing at a time inside each function call.  (`print` lets you separate things with commas, but `write` doesn't let you do this.  You can combine pieces with string concatenation, however).
    
    
* The `write` function works a little differently than other functions, in that you must preface the `write` with the file object from when you opened the file.

````python
# Example with writing numbers to a file.

# Code along with me to help fix the errors.

def main():
    outfile = open("numbers.txt", "w")  # open the file
    
    while True:
        num = int(input("Give me a number: "))
        
        if num == 0:  # end the loop if we type zero
            break
        
        outfile.write(num)
    
    # close the file
    outfile.close()
    
main()
````

### Takeaway:

* When writing numbers to a file, convert them to strings using the `str` function first.
