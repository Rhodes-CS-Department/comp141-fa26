__Two-dimensional list:__ a list that contains other lists as its elements
* Also known as a **nested list**.
* Common to think of two-dimensional lists as having rows and columns.
* Useful for working with data that naturally has two dimensions.
* To process data in a 2d list, you need to use **two indices**.
* Typically use nested loops to process.

## Example

`grid = [[1, 3, 5, 7], [2, 4, 6, 8], [5, 10, 15, 20]]`

Access individual elements by using two sets of square brackets: row, then column -> <font size="+2"><tt><font color=green>list_variable</font>[row][col]</tt></font>



Be careful: the row (how far down) comes first, and the column (how far across) comes second. That's the opposite of the (x, y) order from our graphics programs, where x (how far across) came first.

````python
grid = [[1, 3, 5, 7], [2, 4, 6, 8], [5, 10, 15, 20]]

# Outputs the individual element in grid at row index 2, column index 1
print("grid[2][1] is", grid[2][1])

# Notice that swapping the two indices gives a different item:
print("grid[1][2] is", grid[1][2])
````

You can also get a whole row by using one set of brackets, or the entire 2d list
by using no brackets.

There's no way to get a whole column with one set of brackets, though. To get the items in a column, you need a loop.

````python
grid = [[1, 3, 5, 7], [2, 4, 6, 8], [5, 10, 15, 20]]

print("row 1:", grid[1])
print("whole thing:", grid)
````

## Computing the Number of Rows/Columns in a 2d List

* To calculate the number of rows, use `len(grid)`
* To calculate the number of columns, use `len(grid[0])`

* We will assume in this class that all rows have the same number of columns.  Therefore, `grid[0]` (the first row) has the same number of elements (columns) as `grid[1]` and `grid[2]` and every other row.  

````python
grid = [[1, 3, 5, 7], [2, 4, 6, 8], [5, 10, 15, 20]]

num_rows = len(grid)
print(num_rows)

num_cols = len(grid[0])
print(num_cols)
````

## Two "standard" loops over 2d lists:

```python
for row in range(0, len(grid)):
    for col in range(0, len(grid[0])):
        <do something with grid[row][col]>
```

```python
for col in range(0, len(grid[0])):
    for row in range(0, len(grid)):
        <do something with grid[row][col]>
```

You will recognize this idea from when we learned about nested loops!

Notice that both loops still use `grid[row][col]` to get an item. Only the order of the loops changes. The row index always goes in the first set of brackets.

The particular loop style you choose usually doesn't matter (though sometimes it does).  It just specifies if you want Python to loop over the list
row-by-row or column-by-column.

## Loop patterns/algorithms

- Most patterns/algorithms we've seen for lists can also be applied to 2d lists:
    - Counting the number of times something happens in a list (e.g., number of odd numbers, number of strings starting with a vowel, etc.)
    - Looking at all the items in a list and doing some sort of computation with them (summing, averaging, finding the max/min, etc.)
    - Changing every item in a list into something else (capitalize the first letter of every string, add 1 to each number, etc.)

Except now that we have two dimensions to play with (rows and columns), this opens up more possibilities for interesting algorithms!

- Counting the number of times something happens in a certain row or column.
- Moving things around in the 2d list.
- Many others!


````python
# Example of counting:

# Returns a number. Doesn't change grid.
def count_odds(grid):
    count = 0
    for row in range(0, len(grid)):
        for col in range(0, len(grid[0])):
            if grid[row][col] % 2 == 1:
                count += 1
    return count

# Test case:
grid = [[5, 6, 2, 7],
        [0, 2, 1, 3],
        [9, 5, 8, 4]]
            
print(count_odds(grid))
````

````python
# Example of changing each item:

# Modifies grid. Doesn't return anything.
def double(grid):
    for row in range(0, len(grid)):
        for col in range(0, len(grid[0])):
            grid[row][col] *= 2

# Notice that when you MODIFY items in a 2d list, you don't need to return anything from the function.

# Test case:
grid = [[5, 6, 2, 7],
        [0, 2, 1, 3],
        [9, 5, 8, 4]]
            
double(grid)
print(grid)
````
