# Nested loops with Farmer Phil



Farmer Phil has some cows in his field he needs to feed.  He notices they are standing all in a line.  He wants to program a robot to feed all of the cows, so he imagines that all the cows are standing
along the x-axis of the Cartesian plane, and then he labels each cow with its position:



Now he can easily walk down the row and feed all the cows:

````python
# Feed all the cows in a line:

def main():
    for ctr in range(0, 5, 1):
        print("Feeding cow", ctr)
        
main()
````

**Notice how each cow can be identified by one number.**  That number here is represented here by the `ctr`
counter variable.

***

Later on, Farmer Phil buys some more cows and they can't all fit in a line anymore.  He
rearranges them in a grid:



But now each cow needs **two** numbers to identify it, a **row number**
and a **column number**.  Phil decides to copy his loop to feed all the cows in each row.

````python
# Feed all the cows in a grid:

def main():
    # Feed the first row:
    print("Feeding all cows in row 0")
    for ctr in range(0, 5, 1):
        print("  Feeding cow", ctr)
        
    # Feed the second row:
    print("Feeding all cows in row 1")
    for ctr in range(0, 5, 1):
        print("  Feeding cow", ctr)
        
    # Feed the third row:
    print("Feeding all cows in row 2")
    for ctr in range(0, 5, 1):
        print("  Feeding cow", ctr)
        
main()
````

***

Farmer Phil realizes that if he keeps adding more rows, he will have to copy and paste
his loop over and over.  There's got to be a better way.

Can we make the row part into a loop too?

````python
# Feed all the cows in a grid:

def main():
    for row in range(0, 3, 1):
        print("Feeding all cows in row", row)
        for col in range(0, 5, 1):
            print("  Feeding cow", col)
        
main()
````

````python
# Feed all the cows in a grid, labeled with coordinates:

def main():
    for row in range(0, 3, 1):
        print("Feeding all cows in row", row)
        for col in range(0, 5, 1):
            print("  Feeding cow", row, col)
        
main()
````



This is called a **nested loop:** a loop that is inside another loop.
* The key idea of nested loops is that you have an "outer" loop and an "inner" loop.
* The inner loop goes through **all** of its *iterations* (cycles/loops) for every **one** iteration of the outer loop.
* Therefore, the inner loop completes its iterations faster than the outer loop, similarly to an old-style mechanical odometer in your car.



Nested loops are most often used when you need **two** (or more) counter variables, each
one operating independently of the others.

This often corresponds to looping/iterating over some sort of two-dimensional structure,
though there are plenty of situations where nested loops are used without a structure
like this.

## A small note:

Nested loops should be used when the two counter variables operate independently. 
This means that one counter variable is allowed to take on a *range* of values while
the other one stays the same.

**The easiest way to explain this is that a nested loop should be used when one counter
variable changes while the other does not change.**  This is because a nested loop always
"freezes" the outer loop counter variable while the inner one changes.

There are other situations where this does not apply; where you have two counter variables
and they both change simultaneously (maybe in different patterns).

For instance, if you want to generate this pattern:
```
1 5
2 4
3 3
4 2
5 1
```
this would not use nested loops.

Similarly with this:
```
1 2
2 4
3 6
4 8
5 10
```

````python
# Two examples from above:

def main():
    # Example 1
    for ctr in range(1, 6, 1):
        print(ctr, 6 - ctr)
        
    print()
        
    # Another way to do example 1
    ctr2 = 5                    # second counter variable
    for ctr in range(1, 6, 1):  # ctr is controlled automatically
        print(ctr, ctr2)
        ctr2 -= 1               # manually decrement ctr2
        
    print()
        
    # Example 2
    for ctr in range(1, 6, 1):
        print(ctr, 2 * ctr)
        
    print()
        
    # Another way to do example 2
    ctr2 = 2                    # second counter variable
    for ctr in range(1, 6, 1):  # ctr is controlled automatically
        print(ctr, ctr2)
        ctr2 += 2               # manually increment ctr2
        
        
main()
````
