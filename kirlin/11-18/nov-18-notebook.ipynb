On Monday, we used nested loops to look at every item in a 2d list. Today we'll look at just **one row** or just **one column**. You'll need this on Friday, when you check for three in a row in tic-tac-toe.

Here's our grid from Monday again, with the indices of each item:

`grid = [[1, 3, 5, 7], [2, 4, 6, 8], [5, 10, 15, 20]]`



## Looping over one row

Look at the indices in row 1: `grid[1][0]`, `grid[1][1]`, `grid[1][2]`, `grid[1][3]`. The row index is always 1. Only the column index changes.

So to loop over one row, you only need **one** loop, over the columns. If `r` is the row you want:

```python
for col in range(0, len(grid[0])):
    <do something with grid[r][col]>
```

The loop goes across the columns, so it uses `len(grid[0])`, the number of columns.

````python
# Example of looping over one row:

# Prints the items in row r, one per line. Doesn't change grid, and doesn't return anything.
def print_row(grid, r):
    for col in range(0, len(grid[0])):
        print(grid[r][col])

# Test case:
grid = [[1, 3, 5, 7],
        [2, 4, 6, 8],
        [5, 10, 15, 20]]

print_row(grid, 1)   # should print 2, 4, 6, and 8 on separate lines
````

## Looping over one column

Now look at the indices in column 2: `grid[0][2]`, `grid[1][2]`, `grid[2][2]`. The column index is always 2. Only the row index changes.

So to loop over one column, you only need one loop, over the rows. If `c` is the column you want:

```python
for row in range(0, len(grid)):
    <do something with grid[row][c]>
```

The loop goes down the rows, so it uses `len(grid)`, the number of rows. The row index still goes in the first set of brackets.

````python
# Example of looping over one column:

# Prints the items in column c, one per line. Doesn't change grid, and doesn't return anything.
def print_col(grid, c):
    for row in range(0, len(grid)):
        print(grid[row][c])

# Test case:
grid = [[1, 3, 5, 7],
        [2, 4, 6, 8],
        [5, 10, 15, 20]]

print_col(grid, 2)   # should print 5, 6, and 15 on separate lines
````

````python
# One column: c stays the same, and the loop goes down the rows.
from grid_animations import *

grid = [[1, 3, 5, 7], [2, 4, 6, 8], [5, 10, 15, 20]]
animate_one_col(grid, 2)
````

## Which `len`?

In a loop over one row, our loop needs to run once for each **column**, so the `for` loop uses `range(0, len(grid[0]))`.   In a loop over one column, the loop runs once for each **row**, so we use `range(0, len(grid))`.

Don't mix them up!

**Key idea:** use the `len` that matches what your loop goes across. If a grid is square, the wrong `len` happens to work, so test your functions on grids that aren't square.
