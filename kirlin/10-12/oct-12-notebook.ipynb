## Nested loops versus one loop with two counters

If you are trying to solve a problem and you know you need two counter variables, how do you know if you need nested loops?

Suppose you wanted to print this output:

```
1 5
2 4
3 3
4 2
5 1
```

versus

```
1 1
1 2
1 3
2 1
2 2
2 3
3 1
3 2
3 3
```

The difference to focus on is that in the first example, both counter variables are always changing at the same time.  In the second
example, the counter variables do not always change at the same time --- they clearly change at different rates.

Rule of thumb: When you have two counter variables, if they usually change simultaneously, use one loop with two different counter variables.
If the variables change at different rates, use nested loops.

**Explanation**: In nested loops, the outer counter variable is *frozen* while the inner counter variable changes, so this forces the variables
to change at different rates.  With a single loop, this is not the case.


````python
# Example 1

def main():
    going_down = 5
    for going_up in range(1, 6, 1):
        print(going_up, going_down)
        going_up += 1
        going_down -= 1

main()
````

````python
# Example 1, another way

def main():
    going_up = 1
    going_down = 5
    while going_up <= 5:
        print(going_up, going_down)
        going_up += 1
        going_down -= 1

main()
````

````python
# Example 2

def main():
    for ctr1 in range(1, 4, 1):
        for ctr2 in range(1, 4, 1):
            print(ctr1, ctr2)

main()
````

## `break` statement

* The `break` statement ends a loop immediately (`for` loop or `while` loop).

* Common uses:

  * As seen above, can be used to test a condition in the middle of a `while` loop; this is often used to avoid having to "prime" the loop
  or assign dummy values to a variable before the loop starts.  Often used with `while True:`
  
  * Can be used to end a loop at multiple places within the body.
  
  * Can end a loop under "abnormal" conditions --- something happened that wasn't supposed to happen.
  
## `continue` statement

* The `continue` statement ends the current *iteration* of a loop immediately (`for` loop or `while` loop).  In other words,
when `continue` is encountered, the loop immediately returns to the top of the loop and runs again.

* Whereas `break` is used in a lot of different situations, `continue` is less common.  It's main use is to "skip" an iteration of a loop
when invalid input is detected.

````python
# Code along with me!
# Just roll one die over and over.

import random

def main():
    while True:
        curr_roll = random.randint(1, 6)
        print("Roll is", curr_roll)

    print("End of dice rolling.")

main()
````
