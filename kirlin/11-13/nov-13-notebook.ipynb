# Word Monsters

Monsters are marching toward your base, and the only thing that can stop them is a good vocabulary.

Here's how the game works:

* Every monster's body is made of the letters in its name.
* Each turn, you cast a spell by typing a real English word with 3 to 6 letters. Monster names don't count.
* Every letter in your spell gets blasted off **every** monster on the board. For example, casting `bomb` at a `zombie` leaves `zie`.
* A monster with no letters left is defeated.
* After each spell, every monster takes one step closer to your base.
* If you defeat every monster, you win. If a monster reaches your base, you lose.

Today you'll build this game. You'll write the code that uses lists, and I've written the rest, including all the drawing.

## How the game stores its monsters

The game keeps its monsters in two parallel lists:

```python
monsters  = ["zombie", "imp", "ghost", "bat"]
distances = [5, 7, 4, 8]
```

`monsters` holds the names, and `distances` holds how many steps each monster is from your base. Because they're parallel, items at the same index go together, so the zombie is 5 steps away and the ghost is 4 steps away.

* Spells change the names. When a monster loses all its letters, its name becomes `""`, the empty string. That's a defeated monster.
* Distance 0 is your base, and the far edge of the board is 8.
* On the board, the monsters walk in three lanes, but the lanes are only for drawing. None of your functions need them.

Every function you write takes one or both of these lists. When a function takes both, `monsters` always comes first.

## Three kinds of functions

Every function you write is one of these three kinds. Each step tells you which one, in **bold**, and so does the comment above the function.

* Some functions **modify** an existing list. They change the items of the list they're given, and they don't return anything. The game sees the change because it's the same list.
* Some functions **create a new list**. They build a new list and return it. The list they're given stays the same.
* Some functions **return a number**, or `True` or `False`. They don't change any list.
