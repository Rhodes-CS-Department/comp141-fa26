# Boolean Variables in Python

Previously, we have learned about three Python data types: ints, floats, and strings. Recall that a data type is a
piece of information attached to a value or variable that tells the programming language how to interpret that
variable or value:

```
x = 42      # x is an integer (int) variable (because 42 is written without a decimal point)
y = 42.7    # y is a floating-point (float) variable (because 42.7 is written with a decimal point)
z = "hello" # z is a string (str) variable (because it's written in quotes)
```

Python has many other data types to represent other types of information. One important type is the **Boolean**
data type.

There are only two possible values that a Boolean can be: `True` or `False`. This might seem
strange because ints, floats, and strings have potentially an infinite number of possible values, but Booleans
have only two! They are  normally used in situations where you need to represent the truth of something.

To create and use Boolean variables in Python, you use the special Python keywords `True` and `False`
(they must be capitalized). These are the only two values that may be stored in a Boolean variable.

```
game_over = True   # game_over is a Boolean variable holding the value True
ac_broken = False  # ac_broken is a Boolean variable holding the value False
```

Note: it is very tempting to put quotes around the values `True` and `False`; however, this will create string
variables, not Boolean variables! `"True"` and `"False"` are not the same thing as `True` and `False`!

````python
# Run this cell to see how Python evaluates the keyword True
True
````

````python
# Run this cell to see how Python evaluates the string "True"
"True"
````

````python
# Run this cell to see how Python evaluates the comparison 3 < 5
3 < 5
````

That `True` is exactly the same `True` you get by typing the keyword directly. (And `3 > 5` would evaluate
to `False`.) This is an important idea: a comparison like `3 < 5` is not some special kind of thing that only
works inside an if statement. It is an expression that **evaluates to a Boolean value**, in the same way that
`2 + 2` is an expression that evaluates to the int `4`.

You have been writing comparisons like this inside if statements for a few weeks now.
Now you know what they actually produce: one of the two Boolean values.

And because a comparison produces a value, you can store that value in a variable and use it later.

````python
# The result of a comparison can be stored in a variable, just like a number or a string
temperature = 95
is_hot = temperature > 90
print("Is it hot?", is_hot)
````

## What are Boolean variables good for?

Boolean variables can be used in the condition part of an if statement:

```
if game_over:
    print("Thanks for playing!")
```

Note that the following works, but is considered poor style:

```
if game_over == True:                 # the == True part is unnecessary
    print("Thanks for playing!")
```

This idea is handy when there might be many conditions that could end the game, each happening in a
different part of your code. Imagine a game that could be won or lost in many different ways, each requiring a
different section of code. Each section could test the appropriate condition and set the `game_over` variable to
`True` if the user happens to win or lose. 

## Can they be used for other things?

Boolean variables are also commonly used as the return value of a function that is designed to determine
whether something is true or not. For instance, consider a function that is designed to test whether a number
is even or odd. This can be done by dividing the number by 2 and examining the remainder. An even number, 
when divided by 2, has a remainder of 0, whereas an odd number has a remainder of 1. 

In the following cell, the function `is_even` returns `True` when the number it is passed as a parameter is even, and `False` otherwise.

You have actually written this function before. In Lab 4, you wrote a function called `is_even` that returned
the string `"yes"` when a number was even and `"no"` when it was odd. The version below does the same job, but
returns `True` or `False` instead. In a moment you will see why the Boolean version is easier to work with.

Run the cell to see the output of the function.

````python
def is_even(num):
    if num % 2 == 0:
        return True
    else:
        return False

print("Is 42 even?", is_even(42))
print("Is 11 even?", is_even(11))
````

````python
# Now the function can be used elsewhere...
n = int(input("Enter an integer: "))

even = is_even(n) # Capture the Boolean return value.

if even: # Test the value of the variable even.
    print("Your number is even.")
else:
    print("Your number is odd.")
````

````python
# Actually, capturing the return value in a variable is not necessary...
n = int(input("Enter an integer: "))

if is_even(n): # Call is_even and run the test together!
    print("Your number is even.")
else:
    print("Your number is odd.")
````

Notice the line of code: `if is_even(n):`. When you call a function in the condition part of an if statement,
Python calls the function, then uses the return value of the function call as the if-condition. In other words,
that single line of code acts as if you called the `is_even` function, passing `n` as the argument, then captured
the return value, saved it in a Boolean variable, and used that variable as the condition in the if statement.

This is also why it is better for `is_even` to return `True` or `False` than to return `"yes"` or `"no"`. If the
function returned a string, you could not use it as a condition directly. You would have to write
`if is_even(n) == "yes":`, which is longer, and easy to break by accidentally typing `"Yes"` or `"y"` somewhere.
A function that returns a Boolean can be dropped straight into an if statement.
