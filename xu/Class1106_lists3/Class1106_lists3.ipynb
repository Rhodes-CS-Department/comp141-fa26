# More Lists

## Comparison of two common idioms for strings and lists.

Remember, not **all** problems with strings or lists are solved with loops, but many are.

If your problem requires you to examine each character in a string, or each item in a list, one at a time,
the solution probably requires a loop.

---

Counting the number of times something happens in a **string**:

```python
count = 0
for idx in range(0, len(string)):
    if <test something about string[idx]>:
        count += 1    
return count
```


Counting the number of times something happens in a **list**:

```python
count = 0
for idx in range(0, len(lst)):
    if <test something about lst[idx]>:
        count += 1        
return count
```

---

Creating a new string based on the characters in an existing string:

```python
answer = ""
for idx in range(0, len(string)):
    answer += <some character>  # might be inside an if statement      
return answer
```

Creating a new list based on the items in an existing list:

```python
answer = []
for idx in range(0, len(lst)):
    answer.append(<some item>)  # might be inside an if statement    
return count
```

---

**Modifying** an existing list:

```python
for idx in range(0, len(lst)):
    lst[idx] = <change something about lst[idx]>
```

**Remember**, these are general patterns.  They can be adapted to fit lots of different situations, and some
problems will require you to use these in different ways.

## Parallel lists

**Parallel lists** are where you have two lists with the same length, where each item in one list is 
connected to or associated iwth the corresponding item (at the same index) in the other list.

### Examples

|                | 0            | 1       | 2           | 3         |
|----------------|--------------|---------|-------------|-----------|
| first_names    | "George"     | "John"  | "Thomas"    | "James"   |
| last_names     | "Washington" | "Adams" | "Jefferson" | "Madison" |

```python
# Lists do not have to line up like this, but it's easier to read this way.
first_names = ["George",     "John",  "Thomas",    "James" ]
last_names  = ["Washington", "Adams", "Jefferson", "Madison" ]
```

|          | 0        | 1         | 2         | 3         | 4       |
|----------|----------|-----------|-----------|-----------|---------|
| products  | "apples" | "oranges" | "peaches" | "bananas" | "pears" |
| quantities | 100      | 50        | 75        | 125       | 80      |

```python
products = ["apples", "oranges", "peaches", "bananas", "pears"]
quantities = [100, 50, 75, 125, 80]
```

## Reading lists from files



````python
def main():
    file = open("products.txt", "r")
    
    products = []  # empty lists for products and quantities
    quantities = []  
    
    for line in file:
        prod, quant = line.split(" ")
        quant = int(quant)
        
        products.append(prod)
        quantities.append(quant)
        
    file.close()
        
    print("Products are: ", products)
    print("Quantities are: ", quantities)
    
    # Loop over both lists simultaneously:
    for idx in range(0, len(products)):
        print("Product", idx, "is", products[idx], "and the quantity is", quantities[idx])
    
main()
        
````

````python
# Demonstration of minimum and maximum for lists:

def main():
    # Read the products file.
    
    file = open("products.txt", "r")
    
    products = []
    quantities = []  
    
    for line in file:
        prod, quant = line.split(" ")
        quant = int(quant)
        
        products.append(prod)
        quantities.append(quant)
        
    file.close()
    
    # Find the minimum and maximum quantities.
        
    least_quant = quantities[0]
    most_quant = quantities[0]
    for idx in range(1, len(quantities)):
        if quantities[idx] < least_quant:  # minimum algorithm
            least_quant = quantities[idx]
            
        if quantities[idx] > most_quant:  # maximum algorithm
            most_quant = quantities[idx]
            
    print("Highest quantity:", most_quant)
    print("Lowest quantity:", least_quant)
    
main()
````

**Key idea**: To modify an item in a list, we must know **what index the item is at.**

Therefore, we should save the **index** of the smallest item,
alongside the value of the item itself.

````python
def main():
    file = open("products.txt", "r")
    
    products = []
    quantities = []  
    
    for line in file:
        prod, quant = line.split(" ")
        quant = int(quant)
        
        products.append(prod)
        quantities.append(quant)
        
    file.close()
        
    # Find the smallest quantity.
    least_quant = quantities[0]
    least_quant_idx = 0             # save the index!
    for idx in range(1, len(quantities)):
        if quantities[idx] < least_quant:
            least_quant = quantities[idx]
            least_quant_idx = idx   # save the index!
            
    # Key idea for the loop above: whenever we change least_quant,
    # we also change least_quant_idx, so they are always kept in sync.
            
    print("Lowest quantity:", least_quant)
    print("...is located at index:", least_quant_idx)
    print("...and the product is", products[least_quant_idx])
    # Look at the line above: I can also use the saved index to
    # print the name of the product corresponding to the smallest quantity!
    
    # Buy ten more of the least-quantity product:
    # least_quant += 10  <--- No, doesn't work.
    quantities[least_quant_idx] += 10   # <--- This works!
    
    # Print out the products, after buying more of the least one:
    for idx in range(0, len(products)):
        print("Product", idx, "is", products[idx], "and the quantity is", quantities[idx])
    
    
main()
````
