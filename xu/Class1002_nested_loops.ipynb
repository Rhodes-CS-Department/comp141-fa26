__Nested Loop__:A loop that is inside another loop.
* An inner loop goes through all of its iterations for every single iteration of an outer loop.
* Inner loops complete their iterations faster than outer loops.
* To get the total number of iterations of a nested loop, multiply the number of iterations of all the loops

How many iterations occurred in the nested loop we wrote above?


Syntax of nested loop:
```
while expression:
    while expression:
        statement(s)
    statement(s)
```
```
for iterating_var in sequence:
    for iterating_var in sequence:
        statement(s)
    statement(s)
```
```
for iterating_var in sequence:
    while expression:
        statement(s)
    statement(s)
```

```
while expression:
    for iterating_var in sequence:
        statement(s)
    statement(s)
```


````python
# Nested Loop Example 1
# Times Tables
for i in range(1, 6):
    for j in range(1, 11):
        print(format(i*j, '4d'), end=' ')
    print()
````

````python
# Nested Loop Example 2
# Dependent Loops

for i in range(1, 8, 3):
    for j in range(8, i, -2): # range of j DEPENDS on range of i
        print(j, end=' ')
    
````

````python
#Another dependent nested loop
for i in range(2, 10, 3):
    for j in range(i, 13, 2): # range of j DEPENDS on range of i
        print(j, end=' ')
    print()
````

````python
# Nested Loop Example 3
# Prime Numbers

i = 2
while i < 100: # for every number
    prime = True
    for j in range(2, i): # check if every other number less than it is a factor of it
        if i % j == 0:
            prime = False
            break
    if prime:
        print(i, "is prime")
    i = i + 1
````

````python
# Nested Loops Example 4
# Clock Example

for hrs in range(0, 12):
    for mins in range(0, 60):
        for secs in range(0, 60):
            print (hrs, ":", mins, ":", secs)
````

````python
# Pattern 1: Colorful Circle Grid

from cs1.graphics import *

open_canvas(500, 500)
set_background_color("black")

for row in range(8):
    for col in range(8):
        x = 40 + col * 60
        y = 40 + row * 60

        red = 30 * row
        green = 30 * col
        blue = 255 - 15 * (row + col)

        set_color_rgb(red, blue, green)
        draw_filled_circle(x, y, 20)
````

````python
# Pattern 2: Nested Squares

from cs1.graphics import *

open_canvas(500, 500)
set_background_color("white")

for row in range(5):
    for col in range(5):
        x = 25 + col * 95
        y = 25 + row * 95

        for size in range(80, 10, -15):
            red = 40 * row
            green = 40 * col
            blue = 220

            set_color_rgb(red, blue, green)
            draw_rect(x + (80 - size) / 2,
                      y + (80 - size) / 2,
                      size,
                      size)
````

````python
# Pattern 3: Diagonal Circle Waves

from cs1.graphics import *

open_canvas(600, 400)
set_background_color("black")

for row in range(10):
    for col in range(15):
        x = 20 + col * 40
        y = 20 + row * 40

        radius = 5 + ((row + col) % 4) * 3

        red = (row * 25) % 256
        green = (col * 18) % 256
        blue = ((row + col) * 20) % 256

        set_color_rgb(red, blue, green)
        draw_filled_circle(x, y, radius)
````

## Writing Nested Loops
* Design the outer loop without worrying about what goes inside
* Design what goes inside, ignoring the outer loop.
* Put the pieces together, preserving the nesting.
