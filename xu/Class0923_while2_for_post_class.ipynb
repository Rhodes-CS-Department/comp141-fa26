# While loops to count
While loops are also called **condition-controlled loops** since they check a condition every time.

We have maily used while loops to grab input before, but we can also use them to repeatedly do other stuff. Every time you want to repeat something, we can put it in a while loop. One thing we can do is use them to count to a certain number.

## Example: Fibonacci Numbers

The Fibonacci sequence starts with:

```text
0, 1
```

Every number after that is found by adding the previous two numbers.

For example:

```text
0, 1, 1, 2, 3, 5, 8, 13, ...
```

Write a program that uses a `while` loop to print the first `n` Fibonacci numbers.

For example, if:

```python
n = 7
```

the output should be:

```text
0
1
1
2
3
5
8
```

You will need variables to keep track of:

- the two previous Fibonacci numbers;
- how many numbers have been printed so far.

````python
second_to_last = 0
last = 1
count = 0
until_number = 10

print(0, 0)
print(1, 1)

while count < until_number - 2:
    new_number = second_to_last + last
    print(count + 2, new_number)
    second_to_last = last
    last = new_number
    count += 1
````

## Exercise 1: Counting Flower Petals

You are collecting petals from three different types of flowers:

- 5 daisies
- 4 roses
- 6 sunflowers

Each flower has a random number of petals.

First, import the `random` module:

```python
import random
```

To generate a random integer between two values, use:

```python
random.randint(low, high)
```

For example:

```python
petals = random.randint(10, 20)
```

This gives `petals` a random integer from 10 through 20, including both 10 and 20.

For this exercise:

- each daisy has between 15 and 25 petals;
- each rose has between 20 and 35 petals;
- each sunflower has between 30 and 50 petals.

Use `while` loops to go through all of the flowers and count the total number of petals collected from each type.

At the end, print the number of petals collected from each type of flower. Then print out the total number of petals.

Example output:

```text
Daisy petals: 103
Rose petals: 119
Sunflower petals: 241
Total petals: 463
```

The numbers will be different each time because the number of petals on each flower is random.

````python
import random

# petals = random.randint(10, 20)
# print(petals)

NUM_DAISIES = 5
NUM_ROSES = 4
NUM_SUNFLOWERS = 6
daisies_p = 0
roses_p = 0
sunflower_p = 0
total_p = 0

count = 0
while count < NUM_DAISIES:
    petals = random.randint(10, 20)
    daisies_p += petals
    print("Counted", petals, "daisy petals")
    count += 1

count = 0
while count < NUM_ROSES:
    petals = random.randint(20, 35)
    roses_p += petals
    print("Counted", petals, "rose petals")
    count += 1

count = 0
while count < NUM_SUNFLOWERS:
    petals = random.randint(30, 50)
    sunflower_p += petals
    print("Counted", petals, "sunflower petals")
    count += 1

print("Total daisy petals", daisies_p)
print("Total rose petals", roses_p)
print("Total sunflower petals", sunflower_p)
print("Total petals put together", daisies_p + roses_p + sunflower_p)
````

# The `continue` statement

A `continue` statement allows you to **go back to the top of a loop** early.  Normally, a loop only goes back
to the top (the beginning of the body) when it reaches the end of the body.  This statement allows you to "short-circuit" a loop to make it start over early.

This is often used as another way of doing input validation, because it can be used to "restart" a loop if you receive bad input.



````python
## Only allow odd numbers:

def main():
    num = 1
    total = 0
    while num != 0:
        num = int(input("Give me an odd number: "))

        if num % 2 == 0:          # if the number is even
            print("Bad number!")  # print an error message
            continue              # restart the loop

        total += num      # only add up odd numbers
        
    print("Total =", total)
    
main()
````

````python
## Equivalent way without using continue:

def main():
    num = 1
    total = 0
    while num != 0:
        num = int(input("Give me an odd number: "))

        if num % 2 == 0:            # if the number is even
            print("Bad number!")    # print an error message
        else:                       # if it's not even
            total += num            # add up the (odd) number
            
    print("Total =", total)
    
main()
````

A worker is inspecting 10 packages.

For each package:

- randomly determine whether the package is damaged;
- if the package is damaged, print that it is being skipped and use `continue`;
- otherwise, print that the package is being scanned, weighed, and loaded.

For example:

```text
Package 1
Scanning package...
Weighing package...
Loading package...

Package 2
Damaged package — skipping.

Package 3
Scanning package...
Weighing package...
Loading package...
```

Use:

```python
random.randint(0, 1)
```

to determine whether a package is damaged.

For example, you can let:

```text
0 = not damaged
1 = damaged
```

The purpose of `continue` is to immediately skip the remaining steps for damaged packages and move on to the next package.

````python
count = 0
UNTIL = 10

while count < UNTIL:
    condition = random.randint(0, 1)

    print("Package", count+1)

    count += 1

    if condition == 0:
        print("Damaged package — skipping.\n")
        continue

    print("Scanning package...\nWeighing package...\nLoading package...\n")
    
````

# For Loops

__Count-Controlled loop:__ iterates a specific number of times
* Use a for statement to write count-controlled loop 
* Designed to work with sequence of data items
* Iterates once for each item in the sequence
* General Format: 
```
for variable in [val1, val2, etc]:
	statements
```
* In the format above, variable is a new variable that gets created as part of this for statement - you can use this variable anywhere inside the loop body or after the loop is done iterating


````python
for num in [1, 4, 3, 10, 5]:
    print(num**2)
````

## Using the range function

* The range function simplifies the process of writing a for loop
* range returns an iterable object
* __Iterable:__ contains a sequence of values that can be iterated over

__range characteristics:__
* One argument: used as ending limit 
* Two arguments: starting value and ending limit
* Three arguments: third argument is step value

`range(start=0, end, step=1)` shows that range can take in up to 3 arguments, and if not specified the default value for start is 0, and the default value for step is 1

````python
# When you using a single argument, it describes the end value
#The default value for start is used (0), and the default value for step is used (1)

#equivalent while loop to the below for loop
print("here's the while loop: ")
i = 0
while i < 5:
    print(i)
    i += 1

print("here's the for loop: ")
for i in range(5):
    print(i)

print("here's the for loop: ")
for i in range(20, 25):
    print(i)

print("here's the for loop: ")
for i in range(25, 20, -1):
    print(i)

print("here's the for loop: ")
for i in range(20, 30, 2):
    print(i)

initial = 10
until = 20
step = 2
print("here's the for loop: ")
for i in range(initial, until, step):
    print(i)
````

````python
import random

NUM_DAISIES = 5
daisies_p = 0

count = 0
while count < NUM_DAISIES:
    petals = random.randint(10, 20)
    daisies_p += petals
    print("Counted", petals, "daisy petals")
    count += 1

print("Total daisy petals", daisies_p)

daisies_p = 0

for count in range(NUM_DAISIES):
    petals = random.randint(10, 20)
    daisies_p += petals
    print("Counted", petals, "daisy petals")

print("Total daisy petals", daisies_p)
````
