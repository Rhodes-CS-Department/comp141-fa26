__List:__ a new data type that contains multiple data items

* They are very similar to strings, in that **strings** hold multiple **characters**.

* **Lists** hold multiple **elements** or **items**, where each element or item can be any kind of data
(an int, a float, a string, even another list...)




* Lists exist so programmers can store multiple
related variables together, using a single
name.

* Instead of having ten variables named `num1`,
`num2`, `num3`...etc, often we will have a list variable:

    * `numbers[0]`, `numbers[1]`, `numbers[2]`, ...etc

## Let's build a function that works on lists.

With strings, there is a function called `find`.

This function finds the first place in a string that matches its argument,
and returns the **index** of that match.

`find` returns -1 if the character is not found.

```python
def main():
    string = "computer"
    x = string.find("m")
    print(x)
    
    y = string.find("q")
    print(y)
    
main()
```

However, this function does not exist for lists.

There is a similar function called `index`, but it
is not smart enough to return -1 if the item we're looking for isn't there.


```python
def main():
    lst = [2, 4, 6, 8]
    x = lst.index(4)
    print(x)
    
    y = lst.index(5)
    print(y)
    
main()
```
