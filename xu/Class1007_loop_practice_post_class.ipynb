## Saving previous values in a loop

Suppose we repeatedly roll one die. We want to stop when we roll **snake eyes**: two `1`s in a row.

To recognize two consecutive values, the program must remember both:

- `current_roll`: the value produced during the current iteration,
- `previous_roll`: the value produced during the preceding iteration.

The following warm-up stops when a `6` is rolled. The `break` statement immediately exits the loop.

````python
import random


def main():
    while True:
        current_roll = random.randint(1, 6)
        print("Roll:", current_roll)

        if current_roll == 6:
            break

    print("A 6 was rolled. The loop is over.")


main()
````

### Exercise 1: Two ones in a row

Modify the program so that it stops only after rolling two `1`s in a row.

You will need to:

1. Create `previous_roll` before the loop.
2. Generate `current_roll` inside the loop.
3. Check whether both rolls equal `1`.
4. Update `previous_roll` at the end of every iteration that continues.

Be careful about the order of steps. If you update `previous_roll` too early, it will no longer represent the preceding roll.

````python
import random


def main():
    previous_roll = 0
    count = 0

    while True:
        current_roll = random.randint(1, 6)
        print("Roll:", current_roll)
        print("Previous Roll:", previous_roll)

        if current_roll == 1 and previous_roll == 1:
            break
        
        previous_roll = current_roll
        count += 1

    print("You got a snake eyes! Two 1s in a row! It took", count,"times.")

main()
````

## Using loops to validate user input

Programs often repeat a question until the user provides an acceptable value.

Write a program that repeatedly asks the user for a number:

- If the number is positive, add it to a running total and count it.
- If the number is negative, print an error message and do not include it.
- If the number is `0`, end the loop.

After the loop, print the average of all positive numbers entered. If the user entered no positive numbers, print an appropriate message instead of dividing by zero.

````python
def main():
    total = 0
    count = 0
    inp = 1

    # Complete the input-validation loop.
    while inp != 0:
        inp = int(input("Give a non-negative number. Give 0 to exit."))

        if inp > 0:
            total += inp
            count += 1
        elif inp == 0:
            break
        else:
            print("Error, the number is negative. Please make sure your input is positive.")

    if count == 0:
        print("Error. No valid values were input.")
    else:
        print("The average of these numebrs is", total/count)

main()
````

## Loop practice: Xu Corp quality control

Xu Corp produces 100 canned meals every day. Each meal receives three randomly generated measurements:

- Overall quality must be **greater than 5**.
- Fullness must be **greater than 10 and less than 18**.
- Spiciness must be **between 60 and 90, inclusive**.

Use a `for` loop to inspect all 100 products.

If a product passes every test, print that it is being canned. Otherwise, print every failed requirement. Make the error messages precise. For example:

```text
Product 17: Fullness must be greater than 10 and less than 18, but it was 19.
```

A single product may fail more than one requirement.

````python
import random

NUM_PRODUCTS = 100


def create_product():
    overall_quality = random.randint(0, 10)
    fullness = random.randint(5, 20)
    spiciness = random.randint(50, 100)
    return overall_quality, fullness, spiciness


def main():
    print("Welcome to Xu Corp.")
    print("We inspect the quality of every product.")

    # Use a for loop to inspect NUM_PRODUCTS products.
    # Give products the numbers 1 through 100.
    for i in range(NUM_PRODUCTS):
        overall_quality, fullness, spiciness = create_product()

        pass_inspection = True

        if overall_quality <= 5:
            print("Product", i, ": quality must be greater than 5, but it is", overall_quality)
            pass_inspection = False
        if fullness > 5 and fullness < 18:
            print("Product", i, ": fullness must be greater than 10 and less than 18, but it was", fullness)
            pass_inspection = False
        if spiciness >= 60 and spiciness <= 90:
            print("Product", i, ": spiciness must be between 60 and 90 inclusive, but it was", spiciness)
            pass_inspection = False

        if pass_inspection:
            print("Canning right now...")
        else:
            print("Did not pass inspection...")

main()
````

# Nested Loops

A nested loop is a loop inside another loop. For a 2D pattern:

- the outer loop usually controls the **rows**,
- the inner loop usually controls the **columns**.

The `end=""` argument prevents `print()` from moving to a new line after each symbol. The empty `print()` after the inner loop moves to the next row.

````python
ROWS = 4
COLUMNS = 7

for row in range(ROWS):
    for column in range(COLUMNS):
        print("O", end="")
    print()
````

## 3D stacks of 2D patterns

A warehouse stack can contain several 2D pallets. We can represent this with **three nested loops**:

- the first loop selects the layer,
- the second loop selects a row,
- the third loop selects a column.

The example below prints three identical layers.

````python
LAYERS = 3
ROWS = 4
COLUMNS = 6

for layer in range(LAYERS):
    print("Layer", layer + 1)

    for row in range(ROWS):
        for column in range(COLUMNS):
            print("O", end="")
        print()

    print()
````

```python

```

The outer loop changes the row and therefore the $y$-coordinate. The inner loop changes the column and therefore the $x$-coordinate.
