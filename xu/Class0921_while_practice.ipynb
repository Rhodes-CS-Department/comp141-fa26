## Example: Average Numbers Until `-1`

Write a program that asks the user for numbers over and over.

When the user enters `-1`, stop asking for numbers and print the average of all the numbers entered **before** the `-1`.

Example:

```text
Enter a number: 8
Enter a number: 4
Enter a number: 6
Enter a number: -1
Average: 6.0
```

We need to keep track of:

1. the **total** of the numbers entered;
2. how many numbers have been entered.

Then:

```text
average = total / count
```

````python
# Example: work through this together.
number = float(input("Give me a number. Exit with -1"))
sum0 = 0
count = 0

while number != -1:
    sum0 += number
    count += 1
    number = float(input("Give me a number. Exit with -1"))

print("The average of the numbers is", sum0/count)
````
