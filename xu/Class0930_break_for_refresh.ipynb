# The `break` statement

The `break` statement **immediately exits a loop**.

This is different from `continue`:

- `continue` skips the rest of the current iteration and goes back to the top of the loop.
- `break` stops the loop completely.

General pattern:

```python
while condition:
    # do something

    if something_happens:
        break

    # more code
```


## Example: Finding a Treasure

A player can search up to 10 locations.

At each location, we randomly decide whether the treasure is there.

If the treasure is found, we print a message and use `break` so that we stop searching immediately.


````python
import random

location = 1

while location <= 10:
    print("Searching location", location)

    found = random.randint(0, 1)

    if found == 1:
        print("Treasure found!")
        break

    print("No treasure here.")
    location += 1

print("Search finished.")
````

## `break` vs. `continue`

Consider these two ideas:

### `continue`

```python
if package_is_damaged:
    continue
```

This means:

> Skip the rest of this package and move on to the next one.

### `break`

```python
if treasure_is_found:
    break
```

This means:

> Stop searching completely.

So:

```text
continue  -> next iteration
break     -> leave the loop
```

# `for` Loop Refresh

A `for` loop is useful when we want to repeat something a known number of times.

For example:

```python
for num in [1, 4, 3, 10, 5]:
    print(num)
```

The loop runs once for every value in the sequence.


````python
for num in [1, 4, 3, 10, 5]:
    print(num**2)
````

## `range()`

We often use `range()` with a `for` loop.

### One argument

```python
range(5)
```

produces:

```text
0, 1, 2, 3, 4
```

The ending value is **not included**.


````python
for i in range(5):
    print(i)
````

### Two arguments

```python
range(start, end)
```

Example:

```python
range(3, 8)
```

produces:

```text
3, 4, 5, 6, 7
```


````python
for i in range(3, 8):
    print(i)
````

### Three arguments

```python
range(start, end, step)
```

The third number tells Python how much to change the loop variable each time.

Example:

```python
range(2, 11, 2)
```

produces:

```text
2, 4, 6, 8, 10
```


````python
for i in range(2, 11, 2):
    print(i)
````

### Counting backwards

Use a negative step:

```python
range(5, 0, -1)
```

produces:

```text
5, 4, 3, 2, 1
```


````python
for i in range(5, 0, -1):
    print(i)
````

## `while` Loop vs. `for` Loop

These two loops do the same thing:

### `while`

```python
count = 0

while count < 5:
    print(count)
    count += 1
```

### `for`

```python
for count in range(5):
    print(count)
```

When we already know how many times we want to repeat something, a `for` loop is often simpler.
