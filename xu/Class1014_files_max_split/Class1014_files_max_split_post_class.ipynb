## Today's class: finding the largest or smallest item 

Problem that re-occurs often in CS:

* Finding the largest item in a set of things
where you can only look at each thing once.



Pseudocode for finding the largest number in
a collection of numbers:
<pre>
<font size=+1>
# Make a variable to hold the "largest number I've seen so far":
largest = [smallest possible number that you could ever see]

loop over each number:
    if the current number > largest, then
        largest = current number
</font>
</pre>
After this loop finishes, `largest` will have the largest number in it!

This algorithm above is called the **maximum algorithm**.  It is used to find the largest/biggest/greatest
thing in a collection of things.  The "things" are often numbers or strings, but can be any sort of thing 
that allows to to test whether one thing is "bigger than" another thing.

The algorithm can be reversed to implement the **minimum algorithm**.

```python
# Code with me to find the largest number in randoms.txt

def main():
    file = open("randoms.txt", "r")     # open the file

    largest = 0
    
    for line in file:                  # read one line from the file
        line = line.rstrip()           # "rstrip" the line (remove the newline from the end)
        print(line)                    # print the line (useful for debugging but not required)
        
        num = int(line)
        
        if num > largest:
            largest = num

    file.close()

    print("The largest number is", largest)
    
main()
```

```python
# Code with me to find the largest number in randoms.txt

def main():
    file = open("randoms.txt", "r")     # open the file

    smallest = 100000
    
    for line in file:                  # read one line from the file
        line = line.rstrip()           # "rstrip" the line (remove the newline from the end)
        print(line)                    # print the line (useful for debugging but not required)
        
        num = int(line)
        
        if num < smallest:
            smallest = num

    file.close()

    print("The smallest number is", smallest)
    
main()
```

## Handling multiple items per line in a file

### `split` function

Splits a string into multiple string variables
based on a separator:



```python
# Examples of using split:

name = "George Washington"
first, last = name.split(" ")   # split the name string on a space
print("Split into", first, "and", last)

date = "07/04/1776"
month, day, year = date.split("/")
print("Split into", month, day, year)
```

```python
prices = "110.99=14.99=50.99=58.99"
price1, price2, price3, price4 = prices.split("=")
print("The 4 prices are", price1, price2, price3, price4)
```

## Exercise 1: Draw the Mystery Image

The file `image.txt` contains the information needed to draw a small, pixelated mystery image. Write a program that reads the file and draws the image on the canvas using the `draw_grid_rectangle` function below.

Each line of the file describes **one pixel** and contains five values separated by spaces:

- `grid_column`
- `grid_row`
- `red`
- `green`
- `blue`

The first two values give the pixel’s position in the image grid. The final three values give its color. Each color value is an integer from 0 to 255.

For example, the line `3 5 255 0 0` describes a red pixel at column 3, row 5.

The function `draw_grid_rectangle` is provided in the code. See if you can figure out what it does and how its parameters determine where and how large it draws a rectangle. You should be able to work it out by reading the function.

Using the `draw_grid_rectangle` function, fill in the code below to draw your mystery image:

```python
from cs1.graphics import *

PIXEL_SIZE = 10
NUM_ROWS = 35
NUM_COLUMNS = 35

# What does this function do? Ans: 
def draw_grid_rectangle(column, row, pixel_size):
    x = column * pixel_size
    y = row * pixel_size
    draw_filled_rect(x, y, pixel_size, pixel_size)

def main():
    open_canvas(NUM_COLUMNS * PIXEL_SIZE, NUM_ROWS * PIXEL_SIZE)
    set_background_color_rgb(255, 255, 255)

    file = open("image.txt", "r")

    for line in file:
        line = line.rstrip()
        grid_column, grid_row, red, green, blue = line.split(" ")

        set_color_rgb(int(red), int(green), int(blue))
        draw_grid_rectangle(int(grid_column), int(grid_row), PIXEL_SIZE)
    
    file.close()


main()
```
