It is very common to turn code that computes something about a string into a function.

For example, there are two ways of getting total seconds from a `m:ss` format:

```python
def main():
    time = input("Enter how long it took you to finish the race: ")
    
    minutes = time[0]  # slice
    seconds = time[2:4] # another slice   or time[-2:]
    print("minutes is", minutes, "and seconds is", seconds)
    minutes = int(minutes)
    seconds = int(seconds)
    
    total_seconds = minutes * 60 + seconds

    print("You took", total_seconds, "seconds!")  # fill in the blank with a variable
    
main()
```

Note that this function takes input through an `input` statement, and produces output
through a `print` statement.  We can modify this to turn it into a function.  This will
let us use this code in a more flexible manner.

```python
def total_seconds(time):
    minutes = time[0]  # slice
    seconds = time[2:4] # another slice   or time[-2:]
    
    minutes = int(minutes)
    seconds = int(seconds)
    
    total_seconds = minutes * 60 + seconds

    return total_seconds

# Test cases for this function:
print(total_seconds("1:22"))  # should print 82
print(total_seconds("2:59"))  # should print 179
print(total_seconds("4:30"))  # should print 270
```

Here's another example.  Below is code that counts the number of uppercase characters in a string:

```python
def main():
    count = 0

    my_string = input("Enter a string: ")

    for idx in range(0, len(my_string)):
        ch = my_string[idx]
        if ch.isupper():
            count += 1

    print(count, "of the letters were uppercase.")

main()
```

And here is the function version:

```python
def count_upper(string):
    count = 0

    for idx in range(0, len(string)):
        ch = string[idx]  # make a variable to hold the current character
        if ch.isupper():
            count += 1

    return count

# Test cases for this function:
print(count_upper("BanANA"))  
print(count_upper("abracadabra"))  
print(count_upper("Computer Science 141 is awesome!"))  
```

Or, done a different way:

```python
def count_upper(string):
    count = 0

    for idx in range(0, len(string)):
        if string[idx].isupper():  # test current character directly
            count += 1

    return count

# Test cases for this function:
print(count_upper("BanANA"))  
print(count_upper("abracadabra"))  
print(count_upper("Computer Science 141 is awesome!")) 
```

## Building new strings in a loop

Many problems involving strings require us to build a new string out of an old one.  This is often done using string concatenation inside of a `for` loop.

Example: Above we have a function that counts the number of capital letters in a string.  Suppose we wanted to build a function that **removed** anything from the string that was not a capital letter.  In other words, the string "BanANA" would be turned into just "BANA".  


```python
def keep_capitals(string):
    answer = ""
    
    for idx in range(0, len(string)):
        if string[idx].isupper():
            answer = answer + string[idx]
    
    return answer

# Test cases for this function:
print(keep_capitals("BanANA"))  
print(keep_capitals("abracadabra"))  
print(keep_capitals("Computer Science 141 is awesome!")) 
```

```python
def double_chars(string):
    answer = ""
    
    for idx in range(0, len(string)):
        answer = answer + string[idx] + string[idx]
    
    return answer

print(double_chars("BanANA"))  
print(double_chars("abracadabra"))  
print(double_chars("Computer Science 141 is awesome!")) 
```
